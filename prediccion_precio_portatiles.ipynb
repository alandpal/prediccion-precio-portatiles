```python
import pandas as pd
import numpy as np
import re

# --- 1. FUNCIONES AUXILIARES (PARA V5) ---

def extract_cpu_model(cpu_text):
    """Agrupa los modelos de CPU por gama principal."""
    cpu_text = str(cpu_text)
    if 'Core i7' in cpu_text: return 'Core i7'
    if 'Core i5' in cpu_text: return 'Core i5'
    if 'Core i3' in cpu_text: return 'Core i3'
    if 'Ryzen 7' in cpu_text: return 'Ryzen 7'
    if 'Ryzen 5' in cpu_text: return 'Ryzen 5'
    if 'Ryzen 3' in cpu_text: return 'Ryzen 3'
    if re.search(r'AMD A\d+', cpu_text) or 'A-Series' in cpu_text: return 'AMD A-Series'
    if 'Pentium' in cpu_text: return 'Pentium'
    if 'Celeron' in cpu_text: return 'Celeron'
    if 'Atom' in cpu_text: return 'Atom'
    else: return 'Other' # Agrupamos los raros

def extract_gpu_series(gpu_text):
    """Agrupa las GPUs por serie principal."""
    gpu_text = str(gpu_text)
    if 'GeForce GTX' in gpu_text: return 'Nvidia GTX'
    if 'GeForce RTX' in gpu_text: return 'Nvidia RTX'
    if 'Quadro' in gpu_text: return 'Nvidia Quadro'
    if 'GeForce MX' in gpu_text: return 'Nvidia MX'
    if 'GeForce' in gpu_text: return 'Nvidia GeForce'
    if 'Radeon RX' in gpu_text: return 'AMD Radeon RX'
    if 'Radeon R' in gpu_text: return 'AMD Radeon R'
    if 'Radeon' in gpu_text: return 'AMD Radeon'
    if 'Iris' in gpu_text: return 'Intel Iris'
    if 'Intel HD' in gpu_text or 'Intel UHD' in gpu_text: return 'Intel HD/UHD'
    return 'Other' # Agrupamos los raros

# --- 2. FUNCIÓN PRINCIPAL DE LIMPIEZA (V5) ---

def preparar_datos_v5(data):
    """
    Función de limpieza V5 (Jerárquica/Agrupada)
    Esta es la que decidimos que era mejor.
    """
    df = data.copy()

    # --- Ram ---
    df['Ram'] = df['Ram'].str.replace('GB', '').astype(int)

    # --- Weight ---
    df['Weight'] = df['Weight'].str.replace('kg', '').astype(float).fillna(0)

    # --- ScreenResolution ---
    df['Inches'] = df['Inches'].fillna(0)
    df['ScreenResolution'] = df['ScreenResolution'].fillna('0x0')
    df['Touchscreen'] = df['ScreenResolution'].apply(lambda x: 1 if 'Touchscreen' in x else 0)
    df['IPS_Panel'] = df['ScreenResolution'].apply(lambda x: 1 if 'IPS' in x else 0)
    resolucion = df['ScreenResolution'].str.extract(r'(\d+)x(\d+)').fillna('0')
    df['X_Resolution'] = resolucion[0].astype(int)
    df['Y_Resolution'] = resolucion[1].astype(int)
    with np.errstate(divide='ignore', invalid='ignore'):
        df['PPI'] = (((df['X_Resolution']**2) + (df['Y_Resolution']**2))**0.5) / df['Inches']
    df['PPI'] = df['PPI'].replace([np.inf, -np.inf], 0).fillna(0)
    df.drop(['ScreenResolution', 'Inches', 'X_Resolution', 'Y_Resolution'], axis=1, inplace=True)

    # --- Cpu ---
    df['Cpu'] = df['Cpu'].fillna('')
    df['Cpu_Company'] = df['Cpu'].apply(lambda x: str(x).split()[0])
    df['Cpu_Speed_GHz'] = df['Cpu'].str.extract(r'(\d+\.?\d*)GHz').fillna(0).astype(float)
    df['Cpu_Model'] = df['Cpu'].apply(extract_cpu_model) # Usando Auxiliar 1
    df.drop('Cpu', axis=1, inplace=True)

    # --- Memory ---
    df['Memory'] = df['Memory'].fillna('')
    df['SSD_GB'] = 0; df['HDD_GB'] = 0; df['Flash_GB'] = 0; df['Hybrid_GB'] = 0
    df['SSD_GB'] = df['Memory'].str.extract(r'(\d+\.?\d*)\s*GB\s*SSD').fillna(0).astype(float)
    df['HDD_GB'] = df['Memory'].str.extract(r'(\d+\.?\d*)\s*GB\s*HDD').fillna(0).astype(float)
    df['Flash_GB'] = df['Memory'].str.extract(r'(\d+\.?\d*)\s*GB\s*Flash').fillna(0).astype(float)
    df['Hybrid_GB'] = df['Memory'].str.extract(r'(\d+\.?\d*)\s*GB\s*Hybrid').fillna(0).astype(float)
    df['SSD_TB'] = df['Memory'].str.extract(r'(\d+\.?\d*)\s*TB\s*SSD').fillna(0).astype(float)
    df['HDD_TB'] = df['Memory'].str.extract(r'(\d+\.?\d*)\s*TB\s*HDD').fillna(0).astype(float)
    df['Flash_TB'] = df['Memory'].str.extract(r'(\d+\.?\d*)\s*TB\s*Flash').fillna(0).astype(float)
    df['Hybrid_TB'] = df['Memory'].str.extract(r'(\d+\.?\d*)\s*TB\s*Hybrid').fillna(0).astype(float)
    df['SSD_GB'] = (df['SSD_GB'] + df['SSD_TB'] * 1024).astype(int)
    df['HDD_GB'] = (df['HDD_GB'] + df['HDD_TB'] * 1024).astype(int)
    df['Flash_GB'] = (df['Flash_GB'] + df['Flash_TB'] * 1024).astype(int)
    df['Hybrid_GB'] = (df['Hybrid_GB'] + df['Hybrid_TB'] * 1024).astype(int)
    # Eliminamos Total_Storage_GB (como decidimos)
    df.drop(['Memory', 'SSD_TB', 'HDD_TB', 'Flash_TB', 'Hybrid_TB'], axis=1, inplace=True)

    # --- Gpu ---
    df['Gpu'] = df['Gpu'].fillna('')
    df['Gpu_Company'] = df['Gpu'].apply(lambda x: str(x).split()[0])
    df['Gpu_Series'] = df['Gpu'].apply(extract_gpu_series) # Usando Auxiliar 2
    df['Gpu_Model_Num'] = df['Gpu'].str.extract(r'(\d{3,4})').fillna('Other') # Categórico
    df.drop('Gpu', axis=1, inplace=True)
    
    # --- Columnas irrelevantes ---
    if 'Product' in df.columns: df.drop('Product', axis=1, inplace=True)
    if 'laptop_ID' in df.columns: df.drop('laptop_ID', axis=1, inplace=True)
        
    # --- Rellenar NaNs ---
    num_cols = df.select_dtypes(include=np.number).columns
    df[num_cols] = df[num_cols].fillna(0)
    
    cat_cols_to_fill = ['Cpu_Company', 'Gpu_Company', 'OpSys', 'TypeName']
    for col in cat_cols_to_fill:
        if col in df.columns:
            df[col] = df[col].fillna('Unknown')
            if '' in df[col].unique():
                df[col] = df[col].replace('', 'Unknown')

    return df

# --- 3. APLICACIÓN DE LIMPIEZA Y CODIFICACIÓN (One-Hot Encoding) ---

# Cargar Datos
df_train = pd.read_csv('./data/train.csv')
df_test = pd.read_csv('./data/test.csv')

# Guardar ID de prueba y el Precio (target)
test_ids = df_test['id']
y = df_train['Price_euros'] # Esta será nuestra 'y'

# Quitar 'id' y 'Price_euros' ANTES de limpiar
df_train = df_train.drop(['id', 'Price_euros'], axis=1)
df_test = df_test.drop('id', axis=1)

# Aplicar la limpieza V5
print("Aplicando limpieza V5...")
df_train_limpio = preparar_datos_v5(df_train)
df_test_limpio = preparar_datos_v5(df_test)

# --- ESTE ES EL PASO NUEVO E IMPORTANTE: ONE-HOT ENCODING ---
print("Aplicando One-Hot Encoding...")

# Juntamos train y test temporalmente
# Esto asegura que CUALQUIER categoría (ej. 'Compañía') que solo
# exista en 'test' (o 'train') sea reconocida y se cree la columna.
df_full = pd.concat([df_train_limpio, df_test_limpio], axis=0)

# Aplicamos get_dummies para convertir todo el texto a 0s y 1s
# drop_first=True evita problemas de multicolinealidad (bueno para Reg. Lineal)
df_full_encoded = pd.get_dummies(df_full, drop_first=True)

# Separamos de nuevo
# 'X' (train) y 'X_test_final' (test) ya tienen EXACTAMENTE las mismas columnas
X = df_full_encoded.iloc[:len(df_train_limpio)]
X_test_final = df_full_encoded.iloc[len(df_train_limpio):]

# --- 4. RESULTADO ---
print("\n¡Preparación de datos completa!")
print(f"Forma de X (datos de entrenamiento): {X.shape}")
print(f"Forma de y (target): {y.shape}")
print(f"Forma de X_test_final (datos de prueba): {X_test_final.shape}")

print("\nEjemplo de X (entrenamiento) listo para el modelo:")
print(X.head())
```

```python
import xgboost as xgb
import pandas as pd
from sklearn.model_selection import GridSearchCV

# Asumimos que X, y, X_test_final, y test_ids ya existen y están listos

# ==========================================================
# 1. DEFINIR MODELO Y PARÁMETROS
# ==========================================================

# Usamos el modelo que mejor funcionó (el que optimiza MSE por defecto)
# ya que dio el MAE más bajo (177.40)
xgb_base = xgb.XGBRegressor(
    objective='reg:squarederror',
    random_state=42,
    n_jobs=-1
)

# Definimos el "Grid" (la rejilla) de parámetros a probar.
# ¡CUIDADO! Esto es exhaustivo.
param_grid = {
    'max_depth': [4, 5, 6],           # Profundidad del árbol (alrededor de 5)
    'learning_rate': [0.01, 0.05, 0.1], # Tasa de aprendizaje
    'n_estimators': [500, 1000],        # Número de árboles
    'subsample': [0.8, 1.0],            # Muestra de datos por árbol
    'colsample_bytree': [0.8, 1.0]      # Muestra de características por árbol
}

# Total de combinaciones: 3 * 3 * 2 * 2 * 2 = 72 modelos
# Total de entrenamientos (con cv=3): 72 * 3 = 216 entrenamientos

# ==========================================================
# 2. CONFIGURAR Y EJECUTAR GRIDSEARCHCV
# ==========================================================
grid_search = GridSearchCV(
    estimator=xgb_base,
    param_grid=param_grid,
    scoring='neg_mean_absolute_error', # <--- ¡CLAVE! Optimiza para MAE
    cv=3,                                # 3-fold cross-validation
    verbose=2,                           # Muestra el progreso
    n_jobs=-1                            # Usa todos los núcleos
)

print("Iniciando GridSearchCV... Esto puede tardar MUCHO tiempo.")
# Entrenamos la búsqueda sobre el conjunto de entrenamiento COMPLETO
grid_search.fit(X, y)

# ==========================================================
# 3. MOSTRAR LOS MEJORES RESULTADOS
# ==========================================================
print("\n¡Búsqueda completada!")
print(f"Mejores parámetros encontrados: {grid_search.best_params_}")

# El 'best_score_' será negativo (ej. -175.5).
# Lo multiplicamos por -1 para ver el MAE real en positivo.
best_mae = -grid_search.best_score_

print(f"\nMejor MAE (promedio de CV): {best_mae:.2f}")
print(f"MAE anterior (manual): 177.40")

# ==========================================================
# 4. GENERAR SUMISIÓN CON EL MEJOR MODELO
# ==========================================================

# grid_search automáticamente re-entrena el mejor modelo (best_estimator_)
# con todos los datos, así que podemos usarlo directamente.
best_model = grid_search.best_estimator_

print("\nGenerando predicciones para Kaggle con el modelo optimizado...")
preds_test_final = best_model.predict(X_test_final)

# Crear el archivo
submission_grid = pd.DataFrame({
    'id': test_ids,
    'Price_euros': preds_test_final
})

submission_grid.to_csv('submission_xgb_gridsearch.csv', index=False)
print("¡Archivo 'submission_xgb_gridsearch.csv' creado!")
```
